```python
import pandas as pd
import numpy as np
import re

df = pd.read_excel("ComputationalDesignofHydrolase.xlsx", sheet_name="summary")

# Add all necessary columns
df['substrate_name'] = "4MU-PA"
df['substrate_smiles'] = "CC1=CC(OC2=C1C=CC(OC(CC3=CC=CC=C3)=O)=C2)=O"
df['cofactor_name'] = "Zn"
df['cofactor_smiles'] = "[Zn+2]"
df['substrate_moiety'] = "COC(C)=O"
df['cofactor_moiety'] = "[Zn+2]"
df['Entry'] = df['Design'] + '_'+ df['substrate_name']
df['Entry'] = df['Entry'].astype(str).str.replace(r'\s+', '_', regex=True)
df['Entry'] = df['Entry'].astype(str).str.replace(r';+', '_', regex=True)
df['Entry'] = df['Entry'].str.replace(r'[()]', 'b', regex=True)

df.rename(columns={"Designed Protein Sequence": "Sequence"}, inplace=True)

def clean_sequence(seq: str) -> str:

    if pd.isna(seq):
        return None
    seq = seq.upper()
    seq = re.sub(r'[^ACDEFGHIKLMNPQRSTVWY]', '', seq)  # Keep only standard 20 amino acids
    return seq

df['Sequence'] = df['Sequence'].apply(clean_sequence)
# df.to_pickle("metallohydrolases_input_data.pkl")
```
